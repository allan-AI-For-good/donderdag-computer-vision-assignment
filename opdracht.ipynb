```python
from ultralytics import YOLO

# 1. Load the lightweight pre-trained nano model as a starting point
model = YOLO('yolov8n.pt')

# 2. Train the model on your custom dataset
# Adjust epochs (e.g., 25 or 50) depending on how fast you want it to finish
results = model.train(
    data='data.yaml', 
    epochs=25, 
    imgsz=640, 
    batch=8
)
```

```python
from ultralytics import YOLO

# Load your custom trained model weights
model = YOLO('yolov8n.pt')

# Open webcam and run real-time inference (Press 'q' to quit)
model.predict(source=0, show=True, conf=0.5)
```

```python
from ultralytics import YOLO

# Load your trained model weights
model = YOLO('yolov8n.pt')

# Evaluate on validation data to get metrics
metrics = model.val(data='data.yaml')

print("Precision:", metrics.box.p)
print("Recall:", metrics.box.r)
```
